```python
def categorize_col(df, col, labels):
    col_min  = df[col].min()
    col_25   = df[col].quantile(0.25)
    col_50   = df[col].quantile(0.50)
    col_75   = df[col].quantile(0.75)
    col_max  = df[col].max()

    edges = [col_min, col_25, col_50, col_75, col_max]
    df[col] = pd.cut(df[col], bins=edges, labels=labels, include_lowest=True)
    return df
```

```python
labels = ['not_popular', 'below_avg', 'average', 'popular']

if datatest['Vote_Average'].dtype != 'category':
    categorize_col(datatest, 'Vote_Average', labels)

datatest['Vote_Average'].unique()
```

```python
datatest['Vote_Average'] = pd.to_numeric(datatest['Vote_Average'], errors='coerce')
```
